```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(
    r"E:\youtube-data-analytics\INvideos.csv",
    encoding="utf-8"
)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
df_original = df.copy()
df_clean = df.copy()

print("Original rows:", len(df_original))
print("Working rows:", len(df_clean))
print("Backup created successfully!")
```

```python
# Column names ke extra spaces remove karo
df_clean.columns = df_clean.columns.str.strip()

# Data types check karo
df_clean.info()

# Important numeric columns check karo
numeric_cols = ["views", "likes", "comment_count"]

for col in numeric_cols:
    if col in df_clean.columns:
        print(col, "->", df_clean[col].dtype)
```

```python
# Duplicate rows count
duplicate_count = df_clean.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
df_clean = df_clean.drop_duplicates().copy()

print("Rows after removing exact duplicates:", len(df_clean))
```

```python
missing = df_clean.isnull().sum()

print(missing[missing > 0].sort_values(ascending=False))
```

```python
# Missing values ka percentage
missing_percent = (
    df_clean.isnull().mean() * 100
).sort_values(ascending=False)

display(missing_percent[missing_percent > 0].round(2))
```

```python
for col in ["views", "likes", "comment_count"]:
    if col in df_clean.columns:
        print(f"\n{col}")
        print("Negative values:", (df_clean[col] < 0).sum())
        print("Missing values:", df_clean[col].isnull().sum())
```

```python
print("DATA CLEANING REPORT")
print("-" * 30)
print("Original rows:", len(df_original))
print("Cleaned rows:", len(df_clean))
print("Rows removed:", len(df_original) - len(df_clean))
print("Remaining missing values:", df_clean.isnull().sum().sum())
print("Remaining exact duplicates:", df_clean.duplicated().sum())
```

```python
# Top 10 videos by views
top_10_videos = (
    df_clean
    .dropna(subset=["views", "title"])
    .sort_values(by="views", ascending=False)
    .head(10)
)

# Display selected columns
display(top_10_videos[["title", "views", "likes", "comment_count"]])
```

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Prepare chart data
chart_data = top_10_videos.sort_values(
    by="views",
    ascending=True
)

# Create horizontal bar chart
plt.figure(figsize=(12, 7))

plt.barh(
    chart_data["title"].str.slice(0, 45),
    chart_data["views"]
)

plt.title("Top 10 YouTube Videos by Views", fontsize=16)
plt.xlabel("Views")
plt.ylabel("Video Title")

# Display view counts with commas
plt.gca().xaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f"{x:,.0f}")
)

plt.tight_layout()
plt.show()
```

```python
# Select important numeric columns
engagement_data = df_clean[
    ["views", "likes", "comment_count"]
].copy()

# Convert columns to numeric where possible
for col in ["views", "likes", "comment_count"]:
    engagement_data[col] = pd.to_numeric(
        engagement_data[col],
        errors="coerce"
    )

# Remove rows missing required values
engagement_data = engagement_data.dropna()

print("Rows available for analysis:", len(engagement_data))
display(engagement_data.head())
```

```python
correlation = engagement_data[
    ["views", "likes", "comment_count"]
].corr()

display(correlation.round(3))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))

plt.scatter(
    engagement_data["views"],
    engagement_data["likes"],
    alpha=0.5
)

plt.title("YouTube Views vs Likes")
plt.xlabel("Views")
plt.ylabel("Likes")
plt.ticklabel_format(style="plain", axis="both")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    engagement_data["views"],
    engagement_data["comment_count"],
    alpha=0.5
)

plt.title("YouTube Views vs Comments")
plt.xlabel("Views")
plt.ylabel("Comments")
plt.ticklabel_format(style="plain", axis="both")
plt.tight_layout()
plt.show()
```

```python
print("Correlation with views:")
print(correlation["views"].drop("views").round(3))

print("\nCorrelation between likes and comments:")
print(round(correlation.loc["likes", "comment_count"], 3))
```

```python
engagement_data = engagement_data.copy()

engagement_data["like_rate_pct"] = (
    engagement_data["likes"]
    .div(engagement_data["views"].where(
        engagement_data["views"] > 0
    ))
    .mul(100)
)

engagement_data["comment_rate_pct"] = (
    engagement_data["comment_count"]
    .div(engagement_data["views"].where(
        engagement_data["views"] > 0
    ))
    .mul(100)
)

display(
    engagement_data[
        ["views", "likes", "comment_count",
         "like_rate_pct", "comment_rate_pct"]
    ].head(10).round(2)
)
```

```python
category_summary = (
    df_clean.groupby("category_id")
    .agg(
        total_videos=("video_id", "count"),
        total_views=("views", "sum"),
        average_views=("views", "mean"),
        average_likes=("likes", "mean"),
        average_comments=("comment_count", "mean")
    )
    .sort_values("average_views", ascending=False)
)

display(category_summary.head(10).round(2))
```

```python
import matplotlib.pyplot as plt

top_categories = category_summary.head(10).sort_values(
    "average_views"
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_categories.index.astype(str),
    top_categories["average_views"]
)

plt.title("Top 10 Categories by Average Views")
plt.xlabel("Average Views")
plt.ylabel("Category ID")
plt.tight_layout()
plt.show()
```

```python
import json

category_file = r"E:\youtube-data-analytics\IN_category_id.json"

with open(category_file, "r", encoding="utf-8") as file:
    category_json = json.load(file)

categories = {}

for item in category_json["items"]:
    category_id = int(item["id"])
    category_name = item["snippet"]["title"]
    categories[category_id] = category_name

print(categories)
```

```python
category_summary = category_summary.copy()

category_summary["category_name"] = (
    category_summary.index.astype(int)
    .map(categories)
    .fillna("Unknown")
)

display(category_summary.head(10))
```

```python
import matplotlib.pyplot as plt

top_categories = (
    category_summary
    .sort_values("average_views", ascending=False)
    .head(10)
    .sort_values("average_views")
)

plt.figure(figsize=(11, 7))

plt.barh(
    top_categories["category_name"],
    top_categories["average_views"]
)

plt.title("Top 10 YouTube Categories by Average Views")
plt.xlabel("Average Views")
plt.ylabel("Category")
plt.tight_layout()
plt.show()
```
